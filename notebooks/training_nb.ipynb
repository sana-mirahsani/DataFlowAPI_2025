# Main script to run all process.

## 1.Library

```python
import sys
import os
import pandas as pd

# Add the src folder to the Python path
sys.path.append(os.path.abspath("../src"))

# Import modules:
from data_utils import load_data, save_data, explore_data, clean_data, create_text_column, clean_text, convert_by_TF_IDF, split_data
```

## 2.Data Preprocessing

```python
# load and explore data
data_path = os.path.abspath("../data/train.csv")

df = load_data(data_path) 
explore_data(df)
```

```python
# 1. Clean data : remove duplicates and Nall values
df_clean = clean_data(df)

# 2. Add text column to df
df_clean = create_text_column(df_clean, ["review_title","review_content"])

# 3. Clean text in text column
df_clean["text"] = df_clean["text"].astype(str).apply(clean_text)

print("Text column after cleaning process")
print(df_clean["text"].iloc[0:5])

# 4. Convert text column to numerical vectors by TF-IDF
X, y = convert_by_TF_IDF(df_clean)

# 5. Split data to train and test set
X_train, X_test, y_train, y_test = split_data(X, y, test_size=0.2, random_state=42)

print("Train and test shapes : X_train, X_test, y_train, y_test")
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```
